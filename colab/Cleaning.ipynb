```python
data = pd.read_csv('rupiah.csv', sep=',')
data.columns = data.columns.str.lower()

data
```

```python
def cleansing(sent):
    # Mengubah kata menjadi huruf kecil semua dengan menggunakan fungsi lower()
    string = sent.lower()
    string = re.sub(r'http\S+', '', string)  # Menghapus link
    string = re.sub(r'@\w+', '', string)     # Menghapus mention
    string = re.sub(r'#\w+', '', string)     # Menghapus hashtag
    string = re.sub(r'\d+', '', string)      # Menghapus angka
    string = re.sub(r'[^\w\s]', '', string)   # Menghapus karakter khusus
    return string
```

```python
data['full_text'] = data.full_text.apply(cleansing)
data = data[['full_text']]
data.head()
```

```python
data.to_csv('/content/rupiah_clean.csv', index=False)
```
